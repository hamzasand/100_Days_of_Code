## 4. Counting Vowels in a given word

```python
vowels = ['a', 'e', 'i', 'o', 'u']
word = "hello"
vowel_count = 0
for char in word:
    if char in vowels:
        vowel_count += 1
print(vowel_count)
```

## 5. Count consonants in a given word

```python
vowel = ['a','e','i','o','u']
word = "programming"
count = 0
for char in word:
    if char not in vowel:
        count += 1
print(count)
```

## 6. Counting the number of occurances of a character in astring

```python
word = "programming"
character = "g"
char_fre = 0
for i in word:
    if i == character:
        char_fre += 1
print(char_fre)
```

## 8. Find Maximum Number in the list

```python
list = [12, 3, 55, 23, 6, 78, 33, 4]
max = list[0]
for num in list:
    if num > max:
        max = num
print(max)
```

## 9. Finding Minimum number in list

```python
list = [12, 3, 55, 23, 6, 78, 33, 4]
min = list[0]
for num in list:
    if num < min:
        min = num
print(min)
```
