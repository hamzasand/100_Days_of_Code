## 4. Counting Vowels in a given word

```python
vowels = ['a', 'e', 'i', 'o', 'u']
word = "hello"
vowel_count = 0
for char in word:
    if char in vowels:
        vowel_count += 1
print(vowel_count)
```

## 5. Count consonants in a given word

```python
vowel = ['a','e','i','o','u']
word = "programming"
count = 0
for char in word:
    if char not in vowel:
        count += 1
print(count)
```

## 6. Counting the number of occurances of a character in astring

```python
word = "programming"
character = "g"
char_fre = 0
for i in word:
    if i == character:
        char_fre += 1
print(char_fre)
```

## 13. Comparing two strings for anagrams

```python
str1 = "Listen"
str2 = "Silent"
str1 = str1.replace(" ", "").lower()
str2 = str2.replace(" ", "").lower()
if sorted(str1) == sorted(str2):
    print(True)
else:
    print(False)
```

## 14. Checking for Palindrome using slicing technique

```python
str1 = "Kayak"
str1 = str1.lower()
rev_str1 = ""
for i in range(len(str1)):
    rev_str1 = str1[i] + rev_str1
if rev_str1 == str1:
    print(True)
else:
    print(False)
```

```python
str1 = "Kayak".lower()

if str1 == str1[::-1]:
    print(True)
else:
    print(False)
```

## 16. Counting Digits, Letters, and Space in String

```python
import re
name = "Python is 1"

dig_count = re.sub("[^0-9]", "", name)
letter_count = re.sub("[^a-zA-Z]", "",name)
space_count = re.sub("[ \s]", "",name)
print(len(dig_count))
print(len(letter_count))
print(len(space_count))
```

## 26. Find factorial 

```python
num = int(input("Enter a number to find factorial:"))
factorial = 1
if num <0:
    print("factorial of negative number does not exist")
elif num == 0:
    print("factorila of 0 is 1")
else:
    for i in range(1,num+1):
        factorial = factorial*i
    print(factorial)
```
